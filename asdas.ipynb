```python
import numpy as np
import matplotlib.pyplot as plt
import random

# Configuração inicial
t = np.linspace(0, 2*np.pi, 1000)

# Parâmetros aleatórios para variação do coração
size_var = 0.5 + random.random()  # Varia o tamanho entre 0.5 e 1.5
shape_var_x = 0.8 + random.random() * 0.4  # Varia a forma em x
shape_var_y = 0.8 + random.random() * 0.4  # Varia a forma em y
wiggle = random.random() * 0.1  

# Equação paramétrica do coração com variações aleatórias
x = 16 * np.sin(t)**3 * shape_var_x
y = (13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)) * shape_var_y

# Aplica variação de tamanho
x = x * size_var
y = y * size_var

# Cores  aleatórias
heart_color = (random.random(), random.random(), random.random())
background_color = (random.random(), random.random(), random.random())

# Cria a figura
fig, ax = plt.subplots(figsize=(8, 8))

# Preenche o fundo
ax.fill_between([-20, 20], -20, 20, color=background_color, alpha=0.8)

# Desenha o coração
ax.fill(x, y, color=heart_color, alpha=0.9)

# Adiciona detalhes - contorno
ax.plot(x, y, color='darkred', linewidth=2, alpha=0.6)

# Configurações do gráfico
ax.set_xlim(-20, 20)
ax.set_ylim(-20, 20)
ax.axis('off')

plt.title('Coração Gerado', pad=20, color='darkred')
plt.tight_layout()
plt.show()
```
